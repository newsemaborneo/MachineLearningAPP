```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('fifaDatasets.csv')

df.sort_values(by=['Points'], ascending=False, inplace=True)
df.head(10)
```

```python
df_grouped = df.groupby('Team').mean().reset_index()
df_grouped.sort_values(by=['Points'], ascending=False, inplace=True)
df_grouped.head(10)
```

```python
# visualisasi
plt.figure(figsize=(10,6))
plt.bar(df_grouped['Team'][:10], df_grouped['Points'][:10], color='blue')
plt.xlabel('Team')
plt.ylabel('Average Points')
plt.title('Top 10 Teams by Average Points')
plt.xticks(rotation=45)
plt.show()
```

```python
# visualisais position
plt.figure(figsize=(10,6))
plt.bar(df_grouped['Team'][:10], df_grouped['Position'][:10], color='green')
plt.xlabel('Teams')
plt.ylabel("Position")
plt.title('Top 10 Positions by Team')
plt.xticks(rotation=45)
plt.show()
```

```python
# visualisais position
plt.figure(figsize=(10,6))
plt.bar(df_grouped['Team'][:10], df_grouped['Games Played'][:10], color='green')
plt.xlabel('Teams')
plt.ylabel('Games Played')
plt.title('Top 10 Games Played by Team')
plt.xticks(rotation=45)
plt.show()
```
